```python
import pandas as pd
from pathlib import Path

proc = pd.read_csv(
    "procurement_history.csv",
    parse_dates=["created_at", "received_at"]
)

# Basic relationship checks
print("Total rows:", len(proc))
print("Unique purchase orders:", proc["purchase_order_id"].nunique())
print("Unique purchase order items:", proc["purchase_order_item_id"].nunique())
print("Unique goods receipts:", proc["goods_receipt_id"].nunique())
print("Unique ingredients:", proc["ingredient_id"].nunique())
print("Unique suppliers:", proc["vendor_id"].nunique())

print("\nMissing order/receipt identifiers:")
print(proc[
    ["purchase_order_id", "goods_receipt_id",
     "ingredient_id", "vendor_id", "created_at", "received_at"]
].isna().sum())

print("\nPurchase orders with multiple rows:")
print(
    proc.groupby("purchase_order_id")
        .size()
        .value_counts()
        .sort_index()
        .rename_axis("rows_per_order")
        .rename("number_of_orders")
)

print("\nReceipt counts per purchase order:")
print(
    proc.groupby("purchase_order_id")["goods_receipt_id"]
        .nunique()
        .value_counts()
        .sort_index()
        .rename_axis("unique_receipts_per_order")
        .rename("number_of_orders")
)
```

```python

proc = pd.read_csv(
     "procurement_history.csv",
    parse_dates=["created_at", "received_at"]
)

# Keep rows with a receipt and valid dates
valid = proc.dropna(
    subset=["purchase_order_id", "goods_receipt_id",
            "created_at", "received_at"]
).copy()

# Actual elapsed time in days
valid["actual_lead_time_days"] = (
    valid["received_at"].dt.normalize()
    - valid["created_at"].dt.normalize()
).dt.days

# Check how many distinct values occur within each purchase order
checks = valid.groupby("purchase_order_id").agg(
    row_count=("purchase_order_item_id", "size"),
    receipt_count=("goods_receipt_id", "nunique"),
    supplier_count=("vendor_id", "nunique"),
    ingredient_count=("ingredient_id", "nunique"),
    created_date_count=("created_at", lambda x: x.dt.normalize().nunique()),
    received_date_count=("received_at", lambda x: x.dt.normalize().nunique()),
    lead_time_count=("actual_lead_time_days", "nunique")
)

print("Valid rows:", len(valid))
print("\nPurchase orders with multiple receipt IDs:", 
      (checks["receipt_count"] > 1).sum())
print("Purchase orders with multiple suppliers:", 
      (checks["supplier_count"] > 1).sum())
print("Purchase orders with multiple receipt dates:", 
      (checks["received_date_count"] > 1).sum())
print("Purchase orders with multiple lead-time values:", 
      (checks["lead_time_count"] > 1).sum())

print("\nExamples of purchase orders with multiple receipts:")
display(checks[checks["receipt_count"] > 1].head(10))
```

```python

proc = pd.read_csv(
     "procurement_history.csv",
    parse_dates=["created_at", "received_at"]
)

# Keep rows with valid purchase order and receipt information
valid = proc.dropna(
    subset=[
        "purchase_order_id",
        "goods_receipt_id",
        "created_at",
        "received_at"
    ]
).copy()

# Check each purchase-order/receipt combination
receipt_checks = valid.groupby(
    ["purchase_order_id", "goods_receipt_id"]
).agg(
    row_count=("purchase_order_item_id", "size"),
    created_dates=("created_at", lambda x: x.dt.normalize().nunique()),
    received_dates=("received_at", lambda x: x.dt.normalize().nunique()),
    suppliers=("vendor_id", "nunique"),
    ingredients=("receipt_ingredient_id", "nunique")
)

print("Unique PO + receipt combinations:", len(receipt_checks))
print("Combinations with multiple rows:", (receipt_checks["row_count"] > 1).sum())
print("Combinations with conflicting order dates:", (receipt_checks["created_dates"] > 1).sum())
print("Combinations with conflicting receipt dates:", (receipt_checks["received_dates"] > 1).sum())
print("Combinations with multiple suppliers:", (receipt_checks["suppliers"] > 1).sum())

print("\nExamples where a receipt has multiple rows:")
display(receipt_checks[receipt_checks["row_count"] > 1].head(10))
```
